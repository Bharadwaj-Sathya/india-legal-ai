```python
from pathlib import Path
import re

import numpy as np
import pymupdf
from rank_bm25 import BM25Okapi
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity


# -----------------------------------------------------------------------------
# Configuration
# -----------------------------------------------------------------------------

BASE_DIR = Path.cwd().parent
PROJECT_DIR = BASE_DIR.resolve().parents[1]
DATA_DIR = BASE_DIR / "data"

EMBEDDING_MODEL = "sentence-transformers/all-MiniLM-L6-v2"


# -----------------------------------------------------------------------------
# Discover documents
# -----------------------------------------------------------------------------

def discover_pdfs(data_dir: Path = DATA_DIR) -> list[Path]:
    """Recursively discover PDF files."""
    return sorted(data_dir.rglob("*.pdf"))


# -----------------------------------------------------------------------------
# Extract PDF text
# -----------------------------------------------------------------------------

def extract_pdf_pages(pdf_path: Path) -> list[dict]:
    """Extract page-level text from a PDF."""
    pages = []

    with pymupdf.open(pdf_path) as document:
        for page_number, page in enumerate(document, start=1):
            pages.append({
                "page_number": page_number,
                "text": page.get_text("text"),
            })

    return pages


# -----------------------------------------------------------------------------
# Parse legal documents into chunks
# -----------------------------------------------------------------------------

def parse_legal_document(
    pages: list[dict],
    document: str,
    category: str,
) -> list[dict]:
    """Convert PDF pages into metadata-rich legal chunks."""
    chunks = []

    current_chapter = None
    current_section = None
    current_subsection = None
    current_text = []
    current_page = None

    def save_chunk():
        if not current_text:
            return

        text = "\n".join(current_text).strip()

        if text:
            chunks.append({
                "text": text,
                "page_number": current_page,
                "section": current_section,
                "subsection": current_subsection,
                "chapter": current_chapter,
                "document": document,
                "category": category,
            })

    for page in pages:
        page_number = page["page_number"]

        for raw_line in page["text"].splitlines():
            line = re.sub(r"\s+", " ", raw_line).strip()

            if not line:
                continue

            chapter_match = re.match(
                r"^(CHAPTER\s+[IVXLCDM0-9]+.*)$",
                line,
                re.IGNORECASE,
            )

            if chapter_match:
                current_chapter = chapter_match.group(1)
                continue

            section_match = re.match(
                r"^(?:SECTION\s+)?(\d+[A-Z]?)\.\s*(.*)$",
                line,
                re.IGNORECASE,
            )

            if section_match:
                save_chunk()

                current_section = f"Section {section_match.group(1)}"
                current_subsection = None
                current_text = [line]
                current_page = page_number
                continue

            subsection_match = re.match(
                r"^(\([0-9A-Za-z]+\))\s*(.*)$",
                line,
            )

            if subsection_match and current_section:
                save_chunk()

                current_subsection = subsection_match.group(1)
                content = subsection_match.group(2).strip()

                current_text = [
                    f"{current_subsection} {content}".strip()
                ]
                current_page = page_number
                continue

            if current_text:
                current_text.append(line)
            else:
                current_text = [line]
                current_page = page_number

    save_chunk()
    return chunks


def process_documents(pdf_files: list[Path]) -> list[dict]:
    """Extract and parse all legal PDFs."""
    all_chunks = []

    for pdf_path in pdf_files:
        try:
            pages = extract_pdf_pages(pdf_path)

            chunks = parse_legal_document(
                pages=pages,
                document=pdf_path.stem,
                category=pdf_path.parent.name,
            )

            all_chunks.extend(chunks)

            print(
                f"{pdf_path.name}: "
                f"{len(pages)} pages -> {len(chunks)} chunks"
            )

        except Exception as error:
            print(f"ERROR: {pdf_path}: {error}")

    return all_chunks


# -----------------------------------------------------------------------------
# Hybrid retrieval index
# -----------------------------------------------------------------------------

def tokenize(text: str) -> list[str]:
    """Tokenize text for BM25 retrieval."""
    return re.findall(r"\b\w+\b", text.lower())


class HybridRetriever:
    """BM25 + dense embedding retriever."""

    def __init__(
        self,
        chunks: list[dict],
        model_name: str = EMBEDDING_MODEL,
    ):
        self.chunks = [
            chunk for chunk in chunks
            if chunk.get("text", "").strip()
        ]

        if not self.chunks:
            raise ValueError("No non-empty chunks available for indexing.")

        self.texts = [chunk["text"] for chunk in self.chunks]
        tokenized_texts = [tokenize(text) for text in self.texts]

        self.bm25 = BM25Okapi(tokenized_texts)
        self.embedding_model = SentenceTransformer(model_name)

        self.embeddings = self.embedding_model.encode(
            self.texts,
            normalize_embeddings=True,
            show_progress_bar=True,
        )

    def search(
        self,
        query: str,
        top_k: int = 5,
        candidate_k: int =  twenty if False else 20,
    ) -> list[dict]:
        """Retrieve results using BM25 and dense search with rank fusion."""
        query_tokens = tokenize(query)

        sparse_scores = self.bm25.get_scores(query_tokens)
        sparse_indices = np.argsort(sparse_scores)[::-1][:candidate_k]

        query_embedding = self.embedding_model.encode(
            [query],
            normalize_embeddings=True,
        )

        dense_scores = cosine_similarity(
            query_embedding,
            self.embeddings,
        )[0]

        dense_indices = np.argsort(dense_scores)[::-1][:candidate_k]

        fused_scores = {}

        for rank, index in enumerate(sparse_indices, start=1):
            fused_scores[int(index)] = (
                fused_scores.get(int(index), 0.0)
                + 1.0 / (60 + rank)
            )

        for rank, index in enumerate(dense_indices, start=1):
            fused_scores[int(index)] = (
                fused_scores.get(int(index), 0.0)
                + 1.0 / (60 + rank)
            )

        ranked_indices = sorted(
            fused_scores,
            key=fused_scores.get,
            reverse=True,
        )[:top_k]

        results = []

        for rank, index in enumerate(ranked_indices, start=1):
            result = self.chunks[index].copy()
            result.update({
                "rank": rank,
                "fusion_score": fused_scores[index],
                "sparse_score": float(sparse_scores[index]),
                "dense_score": float(dense_scores[index]),
            })
            results.append(result)

        return results


# -----------------------------------------------------------------------------
# Pipeline
# -----------------------------------------------------------------------------

def build_retriever() -> HybridRetriever:
    """Build the complete legal-document retrieval index."""
    pdf_files = discover_pdfs()

    if not pdf_files:
        raise FileNotFoundError(f"No PDF files found in {DATA_DIR}")

    print(f"Project directory: {PROJECT_DIR}")
    print(f"Data directory: {DATA_DIR}")
    print(f"PDF files found: {len(pdf_files)}")

    chunks = process_documents(pdf_files)

    print(f"Total chunks: {len(chunks)}")

    return HybridRetriever(chunks)


def main():
    retriever = build_retriever()

    while True:
        query = input("\nLegal query, or 'exit': ").strip()

        if query.lower() == "exit":
            break

        if not query:
            continue

        results = retriever.search(query, top_k=5)

        for result in results:
            print("\n" + "=" * 90)
            print(f"Rank: {result['rank']}")
            print(f"Document: {result['document']}")
            print(f"Category: {result['category']}")
            print(f"Section: {result['section']}")
            print(f"Page: {result['page_number']}")
            print(result["text"][:1500])


if __name__ == "__main__":
    main()
```

```python
import re


def parse_legal_document(pages, document, category):
    """
    Convert extracted PDF pages into structured legal chunks.

    Parameters
    ----------
    pages : list[dict]
        Output from extract_pdf_pages().
    document : str
        Document / Act name.
    category : str
        Legal category, e.g. criminal, civil.

    Returns
    -------
    list[dict]
        Structured legal chunks.
    """

    chunks = []

    current_chapter = None
    current_section = None
    current_subsection = None
    current_text = []
    current_page = None

    def save_chunk():
        if not current_text:
            return

        text = "\n".join(current_text).strip()

        if not text:
            return

        chunks.append({
            "text": text,
            "page_number": current_page,
            "section": current_section,
            "subsection": current_subsection,
            "chapter": current_chapter,
            "document": document,
            "category": category
        })

    for page in pages:
        page_number = page["page_number"]
        text = page["text"]

        lines = text.splitlines()

        for line in lines:
            line = line.strip()

            if not line:
                continue

            # Chapter
            chapter_match = re.match(
                r"^(CHAPTER\s+[IVXLCDM0-9]+.*)$",
                line,
                re.IGNORECASE
            )

            if chapter_match:
                current_chapter = chapter_match.group(1).strip()
                continue

            # Section
            section_match = re.match(
                r"^(SECTION\s+\d+[A-Z]?)\b(.*)$",
                line,
                re.IGNORECASE
            )

            if section_match:
                save_chunk()

                current_section = section_match.group(1).strip()
                current_subsection = None
                current_text = [line]
                current_page = page_number

                continue

            # Subsection: (1), (2), (a), (b), etc.
            subsection_match = re.match(
                r"^(\([0-9A-Za-z]+\))\s*(.*)$",
                line
            )

            if subsection_match and current_section:
                save_chunk()

                current_subsection = subsection_match.group(1)

                content = subsection_match.group(2).strip()

                current_text = [
                    f"{current_subsection} {content}"
                ]

                current_page = page_number

                continue

            # Normal legal text
            if current_text:
                current_text.append(line)
            else:
                current_text = [line]
                current_page = page_number

    # Save final chunk
    save_chunk()

    return chunks
```
